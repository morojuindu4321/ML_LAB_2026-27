```python
x=np.array([[1,2,3],
            [4,5,6]])
y=np.array([7,8,9])
a=np.add(x,y)
print("Addition : \n",a)
b=np.subtract(x,y)
print("Subtraction : \n",b)
c=np.multiply(x,y)
print("Multiplication : \n",c)
```

```python
array1=np.array([2,4,6,8])
print("Array before appending \n",array1)
newarray=np.append(array1,[1,3,5,7])
print("New Array \n",newarray)
print("Array after appending \n",array1)
```

```python
array1 = np.array([[1,2,3,4], [5,6,7,8]], dtype=np.int64)
print("array before appending \n",array1)
newarray = np.append(array1, [[7], [8]], axis=1)
print("new array \n", newarray)
print("array after appending \n", array1)
```
